```python
# ==========================================
# STEP 2: CREATE CUSTOMER TABLE
# ==========================================

customers = []

account_types = [
    "Savings",
    "Current",
    "Salary",
    "Premium Savings"
]

genders = ["Male", "Female"]

occupations = [
    "Salaried",
    "Business",
    "Self Employed",
    "Government Employee",
    "Student",
    "Professional"
]

for i in range(10000):

    customer_id = f"CU{i+1:06d}"

    first_name = fake.first_name()
    last_name = fake.last_name()

    gender = random.choice(genders)

    date_of_birth = fake.date_of_birth(
        minimum_age=21,
        maximum_age=65
    )

    age = (datetime.now().date() - date_of_birth).days // 365

    branch = branch_df.sample(1).iloc[0]

    account_type = random.choice(account_types)

    account_balance = round(
        np.random.lognormal(mean=10, sigma=1),
        2
    )

    credit_score = random.randint(550, 850)

    occupation = random.choice(occupations)

    annual_income = round(
        np.random.uniform(180000, 2500000),
        2
    )

    join_date = fake.date_between(
        start_date="-10y",
        end_date="today"
    )

    customers.append({
        "Customer_ID": customer_id,
        "First_Name": first_name,
        "Last_Name": last_name,
        "Gender": gender,
        "Date_of_Birth": date_of_birth,
        "Age": age,
        "City": branch["City"],
        "State": branch["State"],
        "Branch_ID": branch["Branch_ID"],
        "Account_Type": account_type,
        "Account_Balance": account_balance,
        "Credit_Score": credit_score,
        "Occupation": occupation,
        "Annual_Income": annual_income,
        "Join_Date": join_date
    })

print("Customer data generated successfully!")
print("Number of customers:", len(customers))
```

```python
# Convert customer list into a Pandas DataFrame

customer_df = pd.DataFrame(customers)

print("Customer DataFrame created successfully!")
print("Rows:", customer_df.shape[0])
print("Columns:", customer_df.shape[1])

customer_df.head()
```

```python
loans = []

loan_types = [
    "Home Loan",
    "Personal Loan",
    "Car Loan",
    "Education Loan",
    "Business Loan",
    "Gold Loan"
]

for i in range(15000):

    loan_id = f"LN{i+1:06d}"

    customer = customer_df.sample(1).iloc[0]

    customer_id = customer["Customer_ID"]
    branch_id = customer["Branch_ID"]

    loan_type = random.choice(loan_types)

    loan_amount = round(
        np.random.uniform(50000, 5000000),
        2
    )

    interest_rate = round(
        np.random.uniform(7, 18),
        2
    )

    loan_term_months = random.choice([
        12, 24, 36, 48, 60, 84, 120, 180, 240
    ])

    application_date = fake.date_between(
        start_date="-5y",
        end_date="today"
    )

    approval_date = application_date + timedelta(
        days=random.randint(1, 30)
    )

    # Calculate realistic default risk
    risk_score = 0

    if customer["Credit_Score"] < 600:
        risk_score += 3
    elif customer["Credit_Score"] < 700:
        risk_score += 1

    if customer["Annual_Income"] < 400000:
        risk_score += 2

    if loan_amount > customer["Annual_Income"] * 2:
        risk_score += 2

    if interest_rate > 14:
        risk_score += 1

    # Convert risk score into default probability
    default_probability = min(
        0.05 + risk_score * 0.08,
        0.80
    )

    default_flag = (
        1 if random.random() < default_probability else 0
    )

    if default_flag == 1:
        loan_status = "Defaulted"
    else:
        loan_status = random.choice([
            "Active",
            "Closed",
            "Pending"
        ])

    loans.append({
        "Loan_ID": loan_id,
        "Customer_ID": customer_id,
        "Branch_ID": branch_id,
        "Loan_Type": loan_type,
        "Loan_Amount": loan_amount,
        "Interest_Rate": interest_rate,
        "Loan_Term_Months": loan_term_months,
        "Loan_Status": loan_status,
        "Application_Date": application_date,
        "Approval_Date": approval_date,
        "Default_Flag": default_flag
    })

print("Loan data generated successfully!")
print("Number of loans:", len(loans))
```

```python
loan_df = pd.DataFrame(loans)

print("Total loans:", len(loan_df))

print("\nDefault Flag:")
print(loan_df["Default_Flag"].value_counts())

print("\nLoan Status:")
print(loan_df["Loan_Status"].value_counts())
```

```python
transactions = []

transaction_types = [
    "Deposit",
    "Withdrawal",
    "Transfer",
    "Payment",
    "UPI",
    "ATM Withdrawal"
]

transaction_modes = [
    "Cash",
    "UPI",
    "Net Banking",
    "Debit Card",
    "Credit Card",
    "ATM"
]

transaction_statuses = [
    "Success",
    "Success",
    "Success",
    "Success",
    "Failed"
]

for i in range(50000):

    transaction_id = f"TX{i+1:06d}"

    # Select a random customer
    customer = customer_df.sample(1).iloc[0]

    customer_id = customer["Customer_ID"]
    branch_id = customer["Branch_ID"]

    transaction_date = fake.date_between(
        start_date="-3y",
        end_date="today"
    )

    transaction_type = random.choice(transaction_types)

    transaction_amount = round(
        np.random.lognormal(mean=8, sigma=1.2),
        2
    )

    transaction_mode = random.choice(transaction_modes)

    transaction_status = random.choice(transaction_statuses)

    transactions.append({
        "Transaction_ID": transaction_id,
        "Customer_ID": customer_id,
        "Branch_ID": branch_id,
        "Transaction_Date": transaction_date,
        "Transaction_Type": transaction_type,
        "Amount": transaction_amount,
        "Transaction_Mode": transaction_mode,
        "Transaction_Status": transaction_status
    })

print("Transaction data generated successfully!")
print("Number of transactions:", len(transactions))
```

```python
transaction_df = pd.DataFrame(transactions)
print("Transaction DataFrame created successfully!")
print("Rows:",transaction_df.shape[0])
print("Columns:",transaction_df.shape[1])
transaction_df.head()
```

```python
payments = []

payment_statuses = [
    "Paid",
    "Paid",
    "Paid",
    "Late",
    "Missed"
]

payment_methods = [
    "UPI",
    "Net Banking",
    "Debit Card",
    "Cash",
    "Auto Debit"
]

for i in range(40000):

    payment_id = f"PM{i+1:06d}"

    # Select a random loan
    loan = loan_df.sample(1).iloc[0]

    loan_id = loan["Loan_ID"]
    customer_id = loan["Customer_ID"]
    branch_id = loan["Branch_ID"]

    payment_date = fake.date_between(
        start_date="-3y",
        end_date="today"
    )

    payment_amount = round(
        np.random.uniform(1000, 100000),
        2
    )

    payment_status = random.choice(payment_statuses)

    payment_method = random.choice(payment_methods)

    payments.append({
        "Payment_ID": payment_id,
        "Loan_ID": loan_id,
        "Customer_ID": customer_id,
        "Branch_ID": branch_id,
        "Payment_Date": payment_date,
        "Payment_Amount": payment_amount,
        "Payment_Status": payment_status,
        "Payment_Method": payment_method
    })

print("Payment data generated successfully!")
print("Number of payments:", len(payments))
```
