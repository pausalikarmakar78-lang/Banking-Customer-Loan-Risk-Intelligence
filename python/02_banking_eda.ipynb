```python
def credit_category(score):
    if score < 600:
        return "Poor"
    elif score < 700:
        return "Fair"
    elif score < 750:
        return "Good"
    else:
        return "Excellent"

customers["Credit_Category"] = customers["Credit_Score"].apply(
    credit_category
)
```

```python
transactions["Transaction_Date"] = pd.to_datetime(
    transactions["Transaction_Date"]
)

transactions["Month"] = transactions["Transaction_Date"].dt.to_period("M")

print("Month column created successfully!")
```

```python
transactions[["Transaction_Date", "Month"]].head()
```

```python
monthly_transactions = (
    transactions
    .groupby("Month")
    .size()
)

monthly_transactions.head()
```

```python
monthly_transactions.plot(
    figsize=(10,5),
    title="Monthly Transaction Trend"
)

plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
monthly_transaction_amount = (
    transactions
    .groupby("Month")["Amount"]
    .sum()
)

monthly_transaction_amount.head()
```

```python
monthly_transaction_amount.plot(
    figsize=(10,5),
    title="Monthly Transaction Amount Trend"
)

plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")
plt.show()
```

# EDA Summary

## Customer Insights
- The dataset contains 10,000 customers.
- Customer demographics were analyzed using gender, age, occupation, city and state.
- Account types and customer balances were analyzed.
- Credit score distribution was examined and customers were categorized by credit quality.

## Loan Insights
- The dataset contains 15,000 loans.
- Loan types and loan statuses were analyzed.
- Loan amount and interest rate distributions were examined.
- Overall loan default rate was calculated.
- Default rates were compared across different loan types.

## Transaction Insights
- The dataset contains 50,000 transactions.
- Transaction types, modes and statuses were analyzed.
- Transaction amount distribution was examined.
- Monthly transaction trends were analyzed.

## Payment Insights
- The dataset contains 40,000 payments.
- Payment status and payment methods were analyzed.
- Payment amount distribution was examined.

## Risk Analysis
- Customer credit score, annual income, loan amount and interest rate were compared with loan default status.
- These features will be useful for the next stage of the project: loan default prediction.
